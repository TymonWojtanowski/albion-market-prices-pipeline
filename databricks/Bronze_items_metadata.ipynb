#Reading from parquet file

```python
df = spark.read.parquet(
    "/Volumes/albion_project/bronze/data/items_metadata.parquet"
)
```

```python
display(df)
```

#Write to bronze table

```python
(
    df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("albion_project.bronze.items")
)
```
